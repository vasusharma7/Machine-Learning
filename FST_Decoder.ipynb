```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

torch.random.manual_seed(0)

model_path = "microsoft/Phi-4-mini-instruct"

model = AutoModelForCausalLM.from_pretrained(
    model_path,
    device_map="auto",
    torch_dtype="auto",
    trust_remote_code=True,
)
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_path, trust_remote_code=True)
```

```python
messages = [
    {"role": "system", "content": "You are a helpful AI assistant."},
    {"role": "user", "content": "Can you provide ways to eat combinations of bananas and dragonfruits?"},
    {"role": "assistant", "content": "Sure! Here are some ways to eat bananas and dragonfruits together: 1. Banana and dragonfruit smoothie: Blend bananas and dragonfruits together with some milk and honey. 2. Banana and dragonfruit salad: Mix sliced bananas and dragonfruits together with some lemon juice and honey."},
    {"role": "user", "content": "What about solving an 2x + 3 = 7 equation?"},
]

pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)

generation_args = {
    "max_new_tokens": 500,
    "return_full_text": False,
    "temperature": 0.0,
    "do_sample": False,
}

output = pipe(messages, **generation_args)
print(output[0]['generated_text'])
```

```python
import torch
import xgrammar as xgr
from transformers import AutoModelForCausalLM, AutoTokenizer

tokenizer_info = xgr.TokenizerInfo.from_huggingface(tokenizer)
compiler = xgr.GrammarCompiler(tokenizer_info)

json_schema = {
    "type": "object",
    "properties": {
        "food_item": {"type": "string"},
        "nutrients": {
            "type": "object",
            "properties": {
                "nutrient": {"type": "string"},
                "quantity": {"type": "integer"}
                }
            }
    },
    "required": ["food_item", "nutrients"]
}
compiled_grammar = compiler.compile_json_schema(json_schema)

matcher = xgr.GrammarMatcher(compiled_grammar)
token_bitmask = xgr.allocate_token_bitmask(1, tokenizer_info.vocab_size)

prompt = "Estimate all the macro nutrients in one butter croissant along with their quantities.\n"
input_ids = tokenizer.encode(prompt, return_tensors="pt")

# Initial mask
matcher.fill_next_token_bitmask(token_bitmask)

generated_ids = input_ids
for _ in range(100):  # max_new_tokens
    with torch.no_grad():
        logits = model(generated_ids).logits[:, -1, :]

    # Apply XGrammar bitmask to logits (infinitely suppresses invalid tokens)
    xgr.apply_token_bitmask_inplace(logits, token_bitmask)

    # Sample next token
    next_token = torch.argmax(logits, dim=-1, keepdim=True)

    # Update grammar state with the chosen token
    matcher.accept_token(next_token.item())

    if matcher.is_terminated():
        break

    # Update bitmask for the next step
    matcher.fill_next_token_bitmask(token_bitmask)


    generated_ids = torch.cat([generated_ids, next_token], dim=-1)

    if next_token.item() == tokenizer.eos_token_id:
        break

new_tokens = generated_ids[0,len(input_ids):]
print(tokenizer.decode(new_tokens))
```

```python
# To Hook Into Hugging Face model calling code

import torch
import xgrammar as xgr
from transformers import LogitsProcessor

class XGrammarLogitsProcessor(LogitsProcessor):
    def __init__(self, matcher, vocab_size):
        self.matcher = matcher
        self.vocab_size = vocab_size
        self.bitmask = xgr.allocate_token_bitmask(1, vocab_size)

    def __call__(self, input_ids, scores):
        self.matcher.fill_next_token_bitmask(self.bitmask)
        xgr.apply_token_bitmask_inplace(scores, self.bitmask)
        return scores

    def update(self, token_id):
        # Advance the grammar state machine
        self.matcher.accept_token(token_id)
```
